```python
import ast
# Function to explode the 'PORTFOLIO' and 'INITIAL_INVESTMENT_STOCKS' columns
def explode_columns(df, portfolio_col, investment_col):
    df[portfolio_col] = df[portfolio_col].apply(eval)  # Convert string representation of list to actual list
    df[investment_col] = df[investment_col].apply(eval)  # Convert string representation of list to actual list
    df = df.explode([portfolio_col, investment_col])
    return df

# Columns to be exploded
portfolio_col = 'PORTFOLIO'
investment_col = 'INITIAL_INVESTMENT_STOCKS'

# Explode the specified columns
df_exploded = explode_columns(df, portfolio_col, investment_col)
```

```python
# Function to extract specific fields from ORG_INVST column
def extract_org_invst_details(org_invst_str):
    org_invst_dict = ast.literal_eval(org_invst_str)
    return pd.Series({
        'EXPECTED_FINAL_PORTFOLIO_VALUE': org_invst_dict.get('Expected Final Portfolio Value'),
        'Median_Final_Portfolio_Value': org_invst_dict.get('Median Final Portfolio Value'),
        'STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE': org_invst_dict.get('Standard Deviation of Final Portfolio Value'),
        'VALUE_AT_RISK': org_invst_dict.get('Value at Risk (VaR 95%)'),
        'Conditional_Value_at_Risk': org_invst_dict.get('Conditional Value at Risk (CVaR 95%)'),
        'PROBABILITY_OF_LOSS': org_invst_dict.get('Probability of Loss'),
        'SHARPE_RATIO': org_invst_dict.get('Sharpe Ratio')
    })

# Apply the function to the ORG_INVST column and join with the exploded DataFrame
org_invst_details = df['ORG_INVST'].apply(extract_org_invst_details)
df_final = df_exploded.join(org_invst_details)

# Select only the required columns
df_selected = df_final[['MEMBER_ID', 'PORTFOLIO', 'MEMBER_NAME', 'MEMBER_DOB', 'MEMBER_EMPLOYMENT', 'MEMBER_GENDER', 'MEMBER_CITY_TOWN', 'MEMBER_STATE', 'MEMBER_CONTACT_VERIFIED', 'INITIAL_INVESTMENT_STOCKS', 'EXPECTED_FINAL_PORTFOLIO_VALUE', 'STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE', 'VALUE_AT_RISK', 'PROBABILITY_OF_LOSS', 'SHARPE_RATIO']]
```

```python
# Write the DataFrame to a CSV file with column names
df_selected.to_csv('ORIGINAL_INVESTMENT_DETAILS.csv', index=False, header=True)
```

```python
# Function to extract specific fields from TEMP_OPTIMIZED column
def extract_temp_opti_details(temp_opti_str):
    temp_optimized_list = ast.literal_eval(temp_opti_str)
    if isinstance(temp_optimized_list, list) and len(temp_optimized_list) > 0:
        temp_optimized_dict = temp_optimized_list[0]  # Assuming the first element is the dictionary we need
        return pd.Series({
            'OPTIMIZED_EXPECTED_FINAL_PORTFOLIO_VALUE': temp_optimized_dict.get('Expected Final Portfolio Value'),
            'OPTIMIZED_MEDIAN_FINAL_PORTFOLIO_VALUE': temp_optimized_dict.get('Median Final Portfolio Value'),
            'OPTIMIZED_STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE': temp_optimized_dict.get('Standard Deviation of Final Portfolio Value'),
            'OPTIMIZED_VALUE_AT_RISK': temp_optimized_dict.get('Value at Risk (VaR 95%'),
            'OPTIMIZED_CONDITIONAL_VALUE_AT_RISK': temp_optimized_dict.get('Conditional Value at Risk (CVaR 95%'),
            'OPTIMIZED_PROBABILITY_OF_LOSS': temp_optimized_dict.get('Probability of Loss'),
            'OPTIMIZED_SHARPE_RATIO': temp_optimized_dict.get('Sharpe Ratio')
        })
    else:
        return pd.Series({
            'OPTIMIZED_EXPECTED_FINAL_PORTFOLIO_VALUE': None,
            'OPTIMIZED_MEDIAN_FINAL_PORTFOLIO_VALUE': None,
            'OPTIMIZED_STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE': None,
            'OPTIMIZED_VALUE_AT_RISK': None,
            'OPTIMIZED_CONDITIONAL_VALUE_AT_RISK': None,
            'OPTIMIZED_PROBABILITY_OF_LOSS': None,
            'OPTIMIZED_SHARPE_RATIO': None
        })

# Apply the function to the TEMP_OPTIMIZED column and join with the exploded DataFrame
temp_opti_details = df['TEMP_OPTIMIZED'].apply(extract_temp_opti_details)
df_final = df_exploded.join(temp_opti_details)

# Select only the required columns
df_selected = df_final[['MEMBER_ID', 'PORTFOLIO', 
                        'OPTIMIZED_EXPECTED_FINAL_PORTFOLIO_VALUE', 
                        'OPTIMIZED_MEDIAN_FINAL_PORTFOLIO_VALUE',
                        'OPTIMIZED_STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE', 
                        'OPTIMIZED_VALUE_AT_RISK', 
                        'OPTIMIZED_CONDITIONAL_VALUE_AT_RISK', 
                        'OPTIMIZED_PROBABILITY_OF_LOSS', 
                        'OPTIMIZED_SHARPE_RATIO']]

# Save the selected columns to a new CSV file
df_selected.to_csv('optimized_portfolio_details.csv', index=False)
```

```python
def load_data(df, tickers: list):
    stock_data = {}
    min_len = []
    for ticker in tickers:
        data = df[df["STOCK"] == ticker]
        stock_data[ticker] = list(data["ADJ_CLOSE"])
        min_len.append(len(data))
    last_len = min(min_len)
    for ticker in tickers:
        stock_data[ticker] = stock_data[ticker][:last_len]
    
    stock_data = pd.DataFrame(stock_data)
    return stock_data
```

```python
def main(config=config):
    
    # Extract configuration parameters
    tickers = config.get('TICKERS', ["MTFS", "ALPA", "OOGGL"])
    start_date = config.get('START_DATE')
    end_date = config.get('END_DATE')
    initial_investment = config.get('INITIAL_INVESTMENT', 1000)
    num_simulations = config.get('NUM_SIMULATIONS', 10000)
    time_horizon = config.get('TIME_HORIZON', 252)
    risk_free_rate = config.get('RISK_FREE_RATE', 0.0)
    custom_weights = config.get('WEIGHTS')
    optimization_config = config.get('optimization', {})
    optimize = optimization_config.get('OPTIMIZE', True)
    balanced = optimization_config.get('BALANCED', False)
    # Load data
    stock_data = load_data(df, tickers)
    
    # Create portfolio
    portfolio = Portfolio(stock_data)
    portfolio.calculate_returns()
    
    # Annualize returns and covariance
    expected_returns = portfolio.returns.mean() * 252
    covariance_matrix = portfolio.returns.cov() * 252

    # return covariance_matrix, expected_returns
    
    # Determine weights
    if optimize:
        optimizer = PortfolioOptimizer(
            expected_returns,
            covariance_matrix,
            risk_free_rate=risk_free_rate
        )
        if balanced:
            optimal_weights = optimizer.minimize_volatility(target_return=expected_returns.mean())
            # print("\nOptimal Balanced Portfolio Weights:")
        else:
            optimal_weights = optimizer.maximize_sharpe_ratio()
            # print("\nOptimal Portfolio Weights to Maximize Sharpe Ratio:")
        display_optimal_weights(stock_data.columns, optimal_weights)
        weights = optimal_weights
    elif custom_weights:
        weights = custom_weights
        # print("\nUsing Custom Weights:")
        display_optimal_weights(stock_data.columns, weights)
    else:
        num_assets = len(expected_returns)
        weights = [1.0 / num_assets] * num_assets
        # print("\nUsing Equal Weights:")
        # display_optimal_weights(stock_data.columns, weights)

    # print(weights)
    # Perform Monte Carlo Simulation
    simulation = MonteCarloSimulation(portfolio.returns, initial_investment, weights)
    all_cumulative_returns, final_portfolio_values = simulation.run_simulation(
        num_simulations, time_horizon
    )
    
    # Analyze Results
    final_insights = get_simulation_insights(final_portfolio_values, initial_investment)
    # print("**************************")
    # for k, v in final_insights.items():
    #     print(k,":", v)
    # print(final_insights)
    return all_cumulative_returns, final_portfolio_values
```

```python
def optimied_sharpe(config=config):
    # Extract configuration parameters
    tickers = config.get('TICKERS', ["MTFS", "ALPA", "OOGGL"])
    start_date = config.get('START_DATE')
    end_date = config.get('END_DATE')
    initial_investment = config.get('INITIAL_INVESTMENT', 1000)
    num_simulations = config.get('NUM_SIMULATIONS', 10000)
    time_horizon = config.get('TIME_HORIZON', 252)
    risk_free_rate = config.get('RISK_FREE_RATE', 0.0)
    custom_weights = config.get('WEIGHTS')
    optimization_config = config.get('optimization', {})
    optimize = optimization_config.get('OPTIMIZE', True)
    balanced = optimization_config.get('BALANCED', False)
    
    # Load data
    stock_data = load_data(df, tickers)
    
    # Create portfolio
    portfolio = Portfolio(stock_data)
    portfolio.calculate_returns()
    
    # Annualize returns and covariance
    expected_returns = portfolio.returns.mean() * 252
    covariance_matrix = portfolio.returns.cov() * 252
    
    # Determine weights
    if optimize:
        optimizer = PortfolioOptimizer(
            expected_returns,
            covariance_matrix,
            risk_free_rate=risk_free_rate
        )
        optimal_weights = optimizer.maximize_sharpe_ratio()
        weights = optimal_weights
    
    simulation = MonteCarloSimulation(portfolio.returns, initial_investment, weights)
    all_cumulative_returns, final_portfolio_values = simulation.run_simulation(
        num_simulations, time_horizon
    )
    
    final_insights = get_simulation_insights(final_portfolio_values, initial_investment)
    return final_insights, weights
```

```python
import random
from tqdm import tqdm
tqdm.pandas()

def modify_config(tickers, initial):
    config["TICKERS"] = tickers
    config["INITIAL_INVESTMENT"] = initial
    r1, r2 = optimied_sharpe(config)
    r2 = r2 * initial
    if len(r2) != len(tickers):
        raise ValueError("Mismatch between number of tickers and expected portfolio values")
    return [r1, r2]
```
