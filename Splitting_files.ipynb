```python
import ast
# Function to explode the 'PORTFOLIO' and 'INITIAL_INVESTMENT_STOCKS' columns
def explode_columns(df, portfolio_col, investment_col):
    df[portfolio_col] = df[portfolio_col].apply(eval)  # Convert string representation of list to actual list
    df[investment_col] = df[investment_col].apply(eval)  # Convert string representation of list to actual list
    df = df.explode([portfolio_col, investment_col])
    return df

# Columns to be exploded
portfolio_col = 'PORTFOLIO'
investment_col = 'INITIAL_INVESTMENT_STOCKS'

# Explode the specified columns
df_exploded = explode_columns(df, portfolio_col, investment_col)
```

```python
# Function to extract specific fields from ORG_INVST column
def extract_org_invst_details(org_invst_str):
    org_invst_dict = ast.literal_eval(org_invst_str)
    return pd.Series({
        'EXPECTED_FINAL_PORTFOLIO_VALUE': org_invst_dict.get('Expected Final Portfolio Value'),
        'Median_Final_Portfolio_Value': org_invst_dict.get('Median Final Portfolio Value'),
        'STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE': org_invst_dict.get('Standard Deviation of Final Portfolio Value'),
        'VALUE_AT_RISK': org_invst_dict.get('Value at Risk (VaR 95%)'),
        'Conditional_Value_at_Risk': org_invst_dict.get('Conditional Value at Risk (CVaR 95%)'),
        'PROBABILITY_OF_LOSS': org_invst_dict.get('Probability of Loss'),
        'SHARPE_RATIO': org_invst_dict.get('Sharpe Ratio')
    })

# Apply the function to the ORG_INVST column and join with the exploded DataFrame
org_invst_details = df['ORG_INVST'].apply(extract_org_invst_details)
df_final = df_exploded.join(org_invst_details)

# Select only the required columns
df_selected = df_final[['MEMBER_ID', 'PORTFOLIO', 'MEMBER_NAME', 'MEMBER_DOB', 'MEMBER_EMPLOYMENT', 'MEMBER_GENDER', 'MEMBER_CITY_TOWN', 'MEMBER_STATE', 'MEMBER_CONTACT_VERIFIED', 'INITIAL_INVESTMENT_STOCKS', 'EXPECTED_FINAL_PORTFOLIO_VALUE', 'STANDARD_DEVIATION_OF_FINAL_PORTFOLIO_VALUE', 'VALUE_AT_RISK', 'PROBABILITY_OF_LOSS', 'SHARPE_RATIO']]
```
